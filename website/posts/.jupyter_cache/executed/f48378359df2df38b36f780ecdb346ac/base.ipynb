```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

import scipy.stats as ss
```

```python
class Vesicle:
    def __init__(self, fill_rate=None, p_release=0.1, quantal_size=1.0):
        self.p_release = p_release
        self._dist = ss.bernoulli(self.p_release)
        self.quantal_size = quantal_size

    def _check_release(self):
        return self._dist.rvs(size=1)

    def activate(self):
        if self._check_release():
            return self.quantal_size
        else:
            return 0.0


class Synapse:
    def __init__(self, n_vesicles=10, p_distribution=None, q_distribution=None):

        self.n_vesicles = n_vesicles
        self.p_distribution = p_distribution
        self.q_distribution = q_distribution

        self.vesicles = self._create_vesicles()

    def _create_vesicles(self):

        ps = self.p_distribution.rvs(size=self.n_vesicles)
        qs = self.q_distribution.rvs(size=self.n_vesicles)

        return [Vesicle(p_release=p, quantal_size=q) for p, q in zip(ps, qs)]
    
    def activate(self):
        return np.sum([v.activate() for v in self.vesicles])
```

```python
synapse = Synapse(
    n_vesicles=100,
    p_distribution=ss.beta(a=5, b=50),
    q_distribution=ss.exponnorm(K=2, loc=10, scale=2)
)
```

```python
runs = 1000

tight_psps = [synapse.activate() for _ in range(runs)]

fig, ax = plt.subplots()
sns.histplot(tight_psps, ax=ax)
```

```python
ks = np.arange(1, 100, 10)

sims = []
runs = 1000
for k in ks:
    synapse = Synapse(
        n_vesicles=100,
        p_distribution=ss.beta(a=5, b=50),
        q_distribution=ss.exponnorm(K=k, loc=10, scale=2)
    )

    sims.append([synapse.activate() for _ in range(runs)])
```

```python
means = []
errors = []
for psp in sims:

    means.append(np.mean(psp))
    errors.append(np.std(psp))
```

```python
fig, ax = plt.subplots()

ax.scatter(means, errors)
```
